Your answer should replace the definition of the variable `bool_statement`.  The value of `bool_statement` should be `True` if Jan. 13th is available, and `False` otherwise.

```python
# This code creates a list of strings:

with open("quiz_data/dates.txt") as f:
    unavailable = [line.strip() for line in f]

print(unavailable, '\n')

proposed_date = 'Jan. 13'
print(f"An interview can be scheduled for {proposed_date}:")



# your code here

bool_statement = False
found = False
for x in unavailable:
    if(x == proposed_date):
        found = True
if not found:
    bool_statement = True



print(bool_statement)
```
